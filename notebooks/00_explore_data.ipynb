# Exploração dos Dados — ENEM 2025


Este notebook apresenta uma análise exploratória do dataset `baseline.parquet` gerado pelo sistema [Diferential-Privacy-Data-Pipeline](https://github.com/L-Repinaldo/Diferential-Privacy-Data-Pipeline-Experiment) utilizado no experimento de classificação.

A análise tem caráter exploratório e busca contextualizar a estrutura
dos dados, a distribuição do target `Q006` e algumas relações entre
as variáveis utilizadas no experimento.

Este notebook não faz parte da execução principal do experimento de
utilidade ou dos ataques de inferência de associação e não contempla
uma tarefa de regressão.

## 1. CONFIGURAÇÃO

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

PARQUET_PATH = PROJECT_ROOT / "src/data/datasets/enem/enem_2025 - v-2026-09-15_00-20-22/baseline.parquet"


PARQUET_PATH 
```

## 2. CARREGAMENTO

```python
import pandas as pd
```

```python
df = pd.read_parquet(PARQUET_PATH)

print(f"Dimensões: {df.shape}")
print(f"Linhas: {df.shape[0]:,}")
print(f"Colunas: {df.shape[1]}")
```

## 3. ESTRUTURA DOS DADOS

```python
display(df.head())

print("\nTipos das variáveis:")
display(df.dtypes)

print("\nValores ausentes:")
missing = (
    df.isna()
      .mean()
      .mul(100)
      .sort_values(ascending=False)
      .to_frame("missing_%")
)

display(missing)
```

## 4. CARDINALIDADE

```python
cardinality = (
    df.nunique(dropna=True)
      .sort_values(ascending=False)
      .to_frame("unique_values")
)

display(cardinality)
```

## 5. Distribuição colunas

```python

for column in df.columns:

    print(f"\nDistribuição da coluna: {column}")

    target_counts = df[column].value_counts(dropna=False)

    target_percent = (
        df[column]
        .value_counts(normalize=True, dropna=False)
        .mul(100)
    )

    target_distribution = pd.DataFrame({
        "count": target_counts,
        "percentage": target_percent.round(2),
    })

    display(target_distribution)
```

## 6. RELAÇÃO Q006 E FAIXA ETÁRIA

```python
print("\nDistribuição de Q006 por TP_FAIXA_ETARIA")

sex_age = pd.crosstab(
    df["TP_FAIXA_ETARIA"],
    df["Q006"],
    normalize="index"
).mul(100)

display(sex_age.round(2))
```

## 7. VARIÁVEIS DEMOGRÁFICAS

```python
demographic_columns = [
    "TP_COR_RACA",
    "TP_NACIONALIDADE",
    "TP_ESTADO_CIVIL",
    "TP_ST_CONCLUSAO",
    "TP_ENSINO",
    "IN_TREINEIRO",
]

for column in demographic_columns:
    print(f"\n{column} x  Q006")

    distribution = pd.crosstab(
        df[column],
        df["Q006"],
        normalize="index"
    ).mul(100)

    display(distribution.round(2))
```
